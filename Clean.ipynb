```python
import pandas as pd

def convert_time(value):
    if pd.isna(value):
        return None

    value = str(value)

    # Already converted
    if ':' in value:
        return value

    value = value.split('.')[0]  # remove .0 if float
    value = value.zfill(4)

    return f"{value[:2]}:{value[2:]}"
```

```python
df['SCHEDULED_DEPARTURE'] = df['SCHEDULED_DEPARTURE'].apply(convert_time)
df['DEPARTURE_TIME'] = df['DEPARTURE_TIME'].apply(convert_time)
```

```python
df['SCHEDULED_ARRIVAL'] = df['SCHEDULED_ARRIVAL'].apply(convert_time)
df['ARRIVAL_TIME'] = df['ARRIVAL_TIME'].apply(convert_time)
```
